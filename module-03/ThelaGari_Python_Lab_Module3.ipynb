## ১.১  list — একাধিক জিনিস এক নামে

এতদিন একটা variable-এ **একটা** জিনিস রাখতেন:


list-এ **অনেকগুলো** রাখা যায়, সারিবদ্ধ — তৃতীয় বন্ধনীর ভেতরে, কমা দিয়ে আলাদা করে।

## ১.২  for loop — একটার পর একটা

```python

```

বাংলায় পড়ুন: **payments-এর প্রতিটা ঘরের জন্য, সেটাকে `p` নাম দিয়ে, নিচের কাজটা করো।**

````python
payments = ["bkash", "nagad", "card", "rocket"]

for p in payments:
    print(p.upper())
````

চারটা লাইন ছাপা হলো, কিন্তু আপনি লিখলেন **দুই লাইন**। চারশোটা হলেও ওই দুই
লাইনই থাকতো। এটাই loop-এর পুরো ব্যাপার।

### ⚠️ ফাঁকা জায়গাই Python-এর ব্যাকরণ

অন্য ভাষায় দ্বিতীয় বন্ধনী দিয়ে বোঝানো হয় কোনটা ভেতরে। Python-এ **ফাঁকা জায়গা**
দিয়ে। ভেতরের লাইন **চার স্পেস** ভেতরে থাকতে হবে।

````python
# ইচ্ছা করে ভুল — ভেতরের লাইনটা ঠেলা হয়নি
code_with_bug = """
for p in ["bkash", "nagad"]:
print(p)
"""

try:
    exec(code_with_bug)
except IndentationError as e:
    print("IndentationError:", e)
    print("→ কোলনের পরের লাইনটা চার স্পেস ভেতরে ঠেলতে হবে")
````

## ১.৩  if / elif / else — শর্ত দিয়ে পথ বাছা

> যদি এমন হয়, তাহলে এটা করো; নইলে ওটা।

````python
status = "cancelled"

if status == "delivered":
    print("হিসাবে ধরবো")
elif status == "cancelled":
    print("বাদ")
else:
    print("দেখতে হবে")
````

### ⚠️ `=` আর `==` এক না

| চিহ্ন | মানে |
|---|---|
| `=` | এটা ওখানে **রাখো** |
| `==` | দুইটা **সমান কি না** মিলিয়ে দেখো |

`if`-এর ভেতরে **সবসময় দুইটা**। ভুলটা একবার দেখে নিন:

````python
try:
    exec('if status = "cancelled":\n    print("x")')
except SyntaxError as e:
    print("SyntaxError:", e.msg)
    print("→ Python নিজেই বলে দিচ্ছে == দিতে হবে")
````

### `elif` কীভাবে কাজ করে

উপর থেকে নিচে মেলায়। **প্রথমটা যেটা মিলে যায়, সেটা চালায়, বাকিগুলো আর দেখেই না।**
তাই ক্রমটা গুরুত্বপূর্ণ।

````python
for amount in [150, 600, 1500]:
    if amount < 500:
        size = "ছোট"
    elif amount < 1000:
        size = "মাঝারি"
    else:
        size = "বড়"
    print(f"৳{amount} → {size}")
````

## ১.৪  শর্ত জোড়া লাগানো — `and`, `or`, `not`, `in`

| শব্দ | মানে |
|---|---|
| `and` | দুইটাই সত্যি হতে হবে |
| `or` | যেকোনো একটা হলেই চলবে |
| `not` | উল্টো করে দেয় |
| `in` | ভেতরে আছে কি না |

````python
status = "delivered"
amount = 800
payment = "nagad"

print(status == "delivered" and amount > 500)
print(payment == "bkash" or payment == "nagad")
print(not status == "cancelled")

# in — তিনটা or লেখার বদলে একটা list
print(payment in ["bkash", "nagad", "rocket"])

# string-এও কাজ করে
print("kash" in "bkash")
````

## ১.৫  `continue` আর `break`

| শব্দ | কী করে |
|---|---|
| `continue` | **এই চক্করটা** ছেড়ে দাও, loop চলতে থাকবে |
| `break` | **পুরো loop** বন্ধ, বাকিগুলো আর দেখবে না |

````python
print("continue — card বাদ:")
for p in payments:
    if p == "card":
        continue
    print("   ", p)

print()
print("break — rocket-এ থেমে যায়:")
for p in payments:
    if p == "rocket":
        break
    print("   ", p)
````

## ১.৬  জমা করা — loop-এর আসল কাজ

নিয়মটা খুব সহজ: **একটা খালি ঘর রাখুন, আর প্রতিবার একটু করে ভরুন।**

````python
amounts = [1240.50, 350.00, 441.75]

total = 0.0        # loop শুরুর আগে, বাইরে
count = 0

for amount in amounts:
    total = total + amount
    count = count + 1

print(total, count)
print(f"গড়: {total / count:.2f}")
````

### ⚠️ আজকের সবচেয়ে বিপজ্জনক ভুল

জমার ঘরটা যদি **loop-এর ভেতরে** লেখেন, প্রতিবার শূন্য হয়ে যাবে। চালিয়ে দেখুন:

````python
amounts = [1240.50, 350.00, 441.75]

total = 0.0
for amount in amounts:
    total = 0.0                # ← ভুল! ভেতরে
    total = total + amount

print("ভুল ফল:", total)
print("ঠিক হতো:", sum(amounts))
print()
print("কোড চললো, কোনো error এলো না — শুধু উত্তরটা ভুল।")
print("এই ধরনের ভুলই বাস্তবে সবচেয়ে ক্ষতিকর।")
````

### সবচেয়ে বড়টা খোঁজা — নিয়ম একই

````python
amounts = [1240.50, 350.00, 441.75, 812.00]

biggest = 0.0
for amount in amounts:
    if amount > biggest:
        biggest = amount

print("সবচেয়ে বড়:", biggest)
````

## ১.৭  `.split()` আর `.join()` — string ↔ list

ক্লাস ৪-এ নামের মাঝের বাড়তি স্পেস সরাতে আমরা `.replace("  ", " ")` ব্যবহার
করেছিলাম। কিন্তু ফাইলে একটা নাম আছে — `mim   begum` — যেখানে **তিনটা** স্পেস।
`replace` ওটা ধরতে পারে না।

````python
name = "mim   begum"

print(repr(name.title().replace("  ", " ")))        # ✗ এখনো দুইটা স্পেস
print(repr(" ".join(name.title().split())))         # ✓ ঠিক
````

### দুই ধাপে কী হচ্ছে, দেখুন

````python
name = "mim   begum"

words = name.title().split()      # যত স্পেসই থাকুক, শব্দ আলাদা করে — একটা list
print(words)

joined = " ".join(words)          # একটা করে স্পেস দিয়ে আবার জোড়া
print(repr(joined))
````

> মাঝখানে একটা **list** — ওটার জন্যই আজ list শেখা দরকার ছিল।

**প্রথমে পুরোটা চালিয়ে দেখুন**, তারপর উপরে ফিরে গিয়ে একেকটা লাইন বদলে দেখুন
কী হয়।

````python
with open("raw_sales.txt", encoding="utf-8") as fh:
    lines = fh.read().strip().split("\n")

print(f"ফাইলে লাইন আছে: {len(lines)}টা")
print()

# জমার ঘর — loop-এর বাইরে
total_amount   = 0.0
good_orders    = 0
cancelled      = 0
biggest_amount = 0.0
biggest_id     = ""

for raw in lines:
    parts = raw.split(";")

    order_id      = parts[0].strip().upper()
    customer_name = " ".join(parts[2].strip().title().split())
    item_name     = parts[3].strip()
    quantity      = int(parts[4].strip())
    amount        = float(parts[5].strip().replace("৳", "").replace(",", ""))
    status        = parts[7].strip().lower()

    if status == "cancelled":
        print(f"সতর্কতা: {order_id} বাতিল — হিসাবে ধরা হলো না")
        cancelled = cancelled + 1
        continue

    print(f"{order_id}  {customer_name:<18} {item_name:<20} {quantity}টা  ৳{amount:.2f}")

    total_amount = total_amount + amount
    good_orders  = good_orders + 1

    if amount > biggest_amount:
        biggest_amount = amount
        biggest_id     = order_id

# সারাংশ — loop-এর বাইরে, বাঁ দিক ঘেঁষে
print()
print("--------- সারাংশ ---------")
print(f"মোট অর্ডার (বাতিল বাদে): {good_orders}টা")
print(f"বাতিল: {cancelled}টা")
print(f"মোট টাকা: ৳{total_amount:.2f}")
print(f"গড় অর্ডার: ৳{total_amount / good_orders:.2f}")
print(f"সবচেয়ে বড়: {biggest_id} — ৳{biggest_amount:.2f}")
````

### ✔ নিজের উত্তর মিলিয়ে নিন

````python
assert good_orders == 9,                    "মোট অর্ডার ৯ হওয়ার কথা"
assert cancelled == 1,                      "বাতিল ১টা হওয়ার কথা"
assert round(total_amount, 2) == 5097.75,   "মোট টাকা ৳5097.75 হওয়ার কথা"
assert biggest_id == "ORD100045",           "সবচেয়ে বড় ORD100045"
print("🎉 সব মিলেছে! পর্ব ১ শেষ।")
````

## ২.৪  `return` — আজকের সবচেয়ে জরুরি জিনিস ⭐

এটাই সবচেয়ে বেশি গুলায়। **`print` দেখায়। `return` ফেরত দেয়।** দুইটা এক না।

````python
def double_print(x):
    print(x * 2)          # শুধু দেখায়

y = double_print(5)
print("y-তে আছে:", y)     # None!
````

দশ ছাপা হলো, কিন্তু `y`-তে কিছু নেই — `None`।
কারণ function পর্দায় দেখিয়ে দিলো, কিন্তু **কিছু ফেরত দিলো না**।

````python
def double_return(x):
    return x * 2          # ফল ফেরত দেয়

y = double_return(5)
print("y-তে আছে:", y)
````

### পার্থক্যটা কোথায় টের পাবেন

`return` করা জিনিস দিয়ে **আরো কাজ করা যায়**:

````python
print(double_return(5) + double_return(3))     # ১০ + ৬ = ১৬

try:
    print(double_print(5) + double_print(3))
except TypeError as e:
    print("TypeError:", e)
    print("→ None দিয়ে যোগ করা যায় না")
````

> **নিয়ম:** যে function **হিসাব করে**, সে `return` দেবে।
> যে function **দেখায়**, সে `print` করবে।
>
> function কিছু `return` না করলে Python নিজে থেকে `None` ফেরত দেয়।

এতক্ষণ সব এক জায়গায়। এবার ভাগ করবো — function গুলো **এক ফাইলে**, ব্যবহার

````python
%%writefile sales_tools.py
"""ThelaGari-র record পরিষ্কার করার যন্ত্রপাতি।"""


def clean_id(text):
    return text.strip().upper()


def clean_name(text):
    return " ".join(text.strip().title().split())


def clean_amount(text):
    text = text.replace("৳", "").replace(",", "").strip()
    return float(text)


def taka(amount):
    return f"৳{amount:.2f}"


def is_cancelled(status):
    return status.strip().lower() == "cancelled"


if __name__ == "__main__":
    print("sales_tools — পরীক্ষা")
    print(clean_id("  ord100045 "))
    print(clean_name("mim   begum"))
    print(clean_amount(" ৳ 1,240.50 "))
````

এবার সেটা import করি:

````python
import sales_tools

print(sales_tools.clean_id("  ord100045 "))
print(sales_tools.clean_name("  RAHIM  uddin "))
print(sales_tools.taka(sales_tools.clean_amount(" ৳ 1,240.50 ")))
````

2. ডাকার সময় **module-এর নাম সামনে বসে** — `sales_tools.taka(...)`

## 🧪 মিনি-ল্যাব ২ — একই কাজ, এবার গোছানো

মিনি-ল্যাব ১-এর **হুবহু একই উত্তর** আসবে। পার্থক্য শুধু — কোডটা এখন
function-এ ভাগ করা, আর পরিষ্কার করার কাজগুলো `sales_tools` module-এ।

````python
import statistics
import sales_tools


def read_lines(path):
    """ফাইল পড়ে লাইনগুলোর list ফেরত দেয়।"""
    with open(path, encoding="utf-8") as fh:
        return fh.read().strip().split("\n")


def summarise(lines):
    """সব record ঘুরে দেখে — টাকার list, বাতিলের সংখ্যা, সবচেয়ে বড়টা।"""
    amounts = []
    cancelled = 0
    biggest_id = ""
    biggest_amount = 0.0

    for raw in lines:
        parts = raw.split(";")

        order_id = sales_tools.clean_id(parts[0])
        name     = sales_tools.clean_name(parts[2])
        item     = parts[3].strip()
        quantity = int(parts[4].strip())
        amount   = sales_tools.clean_amount(parts[5])

        if sales_tools.is_cancelled(parts[7]):
            print(f"সতর্কতা: {order_id} বাতিল — হিসাবে ধরা হলো না")
            cancelled += 1
            continue

        print(f"{order_id}  {name:<18} {item:<20} {quantity}টা  {sales_tools.taka(amount)}")
        amounts.append(amount)

        if amount > biggest_amount:
            biggest_amount = amount
            biggest_id = order_id

    return amounts, cancelled, biggest_id, biggest_amount


def main():
    lines = read_lines("raw_sales.txt")
    print(f"ফাইলে লাইন আছে: {len(lines)}টা")
    print()

    amounts, cancelled, biggest_id, biggest_amount = summarise(lines)

    print()
    print("--------- সারাংশ ---------")
    print(f"মোট অর্ডার (বাতিল বাদে): {len(amounts)}টা")
    print(f"বাতিল: {cancelled}টা")
    print(f"মোট টাকা: {sales_tools.taka(sum(amounts))}")
    print(f"গড় অর্ডার: {sales_tools.taka(statistics.mean(amounts))}")
    print(f"সবচেয়ে বড়: {biggest_id} — {sales_tools.taka(biggest_amount)}")

    return amounts, cancelled, biggest_id


result = main()
````

### ✔ মিলিয়ে নিন

````python
amounts, cancelled, biggest_id = result

assert len(amounts) == 9,                  "৯টা অর্ডার হওয়ার কথা"
assert cancelled == 1,                     "বাতিল ১টা"
assert round(sum(amounts), 2) == 5097.75,  "মোট ৳5097.75"
assert biggest_id == "ORD100045",          "সবচেয়ে বড় ORD100045"
print("🎉 মিনি-ল্যাব ১-এর সাথে হুবহু মিলেছে!")
print()
print("একই উত্তর, কিন্তু কোডটা এখন অন্য কেউ পড়ে বুঝতে পারবে।")
print("আজকের পুরো কথা এটাই।")
````

## 📌 এক নজরে — যা শিখলেন

### পর্ব ১ (ক্লাস ৫)

| জিনিস | এক লাইনে |
|---|---|
| `list` | একাধিক জিনিস এক নামে; গোনা শুরু **শূন্য** থেকে |
| `for` | প্রতিটা ঘরে একবার; ভেতরের লাইন **চার স্পেস** ভেতরে |
| `if/elif/else` | শর্ত দিয়ে পথ বাছা; `=` রাখে, `==` মেলায় |
| `and/or/not/in` | শর্ত জোড়া লাগানো |
| `continue` / `break` | একটা চক্কর বাদ / পুরো loop বন্ধ |
| জমা করা | জমার ঘর **loop-এর বাইরে** |
| `.split()` / `.join()` | string ↔ list |

### পর্ব ২ (ক্লাস ৬)

| জিনিস | এক লাইনে |
|---|---|
| `def` | কাজটাকে নাম দেওয়া; বানানো ≠ ডাকা |
| parameter | খালি ঘর, ডাকার সময় ভরে দেন |
| `return` | **print দেখায়, return ফেরত দেয়** |
| default | না দিলেও চলে; সবসময় শেষে রাখুন |
| scope | ভেতরের variable বাইরে নেই |
| module | প্রতিটা `.py` ফাইলই একটা module |
| `__main__` | নিজে চালালে চলবে, import করলে না |
| standard library | নিজে লেখার আগে দেখে নিন আছে কি না |
